```python
def armstrong_breakdown(num):
    """
    Show detailed breakdown of Armstrong number calculation
    """
    num_str = str(num)
    num_digits = len(num_str)
    
    print(f"\nNumber: {num}")
    print(f"Number of digits: {num_digits}")
    print(f"Calculation: ", end="")
    
    calculations = []
    total = 0
    
    for digit in num_str:
        d = int(digit)
        power = d ** num_digits
        calculations.append(f"{d}^{num_digits}")
        total += power
    
    print(" + ".join(calculations))
    
    # Show step by step
    print("Step by step:")
    for digit in num_str:
        d = int(digit)
        power = d ** num_digits
        print(f"  {d}^{num_digits} = {power}")
    
    print(f"  Sum = {total}")
    print(f"  Is Armstrong? {total == num} (sum {total} {'==' if total == num else '!='} original {num})")
    
    return total == num

# Test with examples
armstrong_breakdown(153)
armstrong_breakdown(370)
armstrong_breakdown(9474)
armstrong_breakdown(100)
```

## Method 2: Detailed Breakdown of Armstrong Numbers

```python
def is_armstrong(num):
    """
    Check if a number is an Armstrong number
    Args: num (int): The number to check
    Returns: bool: True if Armstrong number, False otherwise
    """
    # Convert to string to easily access digits
    num_str = str(num)
    num_digits = len(num_str)
    
    # Calculate sum of digits raised to power of number of digits
    sum_of_powers = sum(int(digit) ** num_digits for digit in num_str)
    
    # Check if it equals the original number
    return sum_of_powers == num

# Test the function
test_numbers = [153, 370, 371, 407, 1634, 9474, 10, 100, 1, 9]

print("Armstrong Number Check:")
print("=" * 50)
for num in test_numbers:
    result = is_armstrong(num)
    num_str = str(num)
    num_digits = len(num_str)
    sum_val = sum(int(digit) ** num_digits for digit in num_str)
    
    print(f"{num:5d} : {str(result):5s} (sum = {sum_val})")
```

# Armstrong Numbers (Narcissistic Numbers)

An Armstrong number (also called a narcissistic number) is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

**Example:**
- 153 is an Armstrong number: 1³ + 5³ + 3³ = 1 + 125 + 27 = 153
- 9474 is an Armstrong number: 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474
- 10 is NOT an Armstrong number: 1¹ + 0¹ = 1 ≠ 10
