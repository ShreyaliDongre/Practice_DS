```python
def armstrong_breakdown(num):
    """Show detailed breakdown of Armstrong number calculation"""
    num_str = str(num)
    num_digits = len(num_str)
    
    print(f"\nNumber: {num}")
    print(f"Number of digits: {num_digits}")
    print(f"Calculation: ", end="")
    
    calculations = []
    total = 0
    
    for digit in num_str:
        d = int(digit)
        power = d ** num_digits
        calculations.append(f"{d}^{num_digits}")
        total += power
    
    print(" + ".join(calculations))
    
    print("Step by step:")
    for digit in num_str:
        d = int(digit)
        power = d ** num_digits
        print(f"  {d}^{num_digits} = {power}")
    
    print(f"  Sum = {total}")
    print(f"  Is Armstrong? {total == num}")
    return total == num

# Example breakdown
armstrong_breakdown(153)
armstrong_breakdown(9474)
```

## Method 2: Detailed Breakdown

```python
def is_armstrong(num):
    """
    Check if a number is an Armstrong number
    Args: num (int): The number to check
    Returns: bool: True if Armstrong number, False otherwise
    """
    num_str = str(num)
    num_digits = len(num_str)
    sum_of_powers = sum(int(digit) ** num_digits for digit in num_str)
    return sum_of_powers == num

# Test the function
test_numbers = [153, 370, 371, 407, 1634, 9474, 10, 100]

print("Armstrong Number Check:")
print("=" * 50)
for num in test_numbers:
    result = is_armstrong(num)
    print(f"{num}: {result}")
```

# Armstrong Numbers (Narcissistic Numbers)

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

**Examples:**
- 153 = 1³ + 5³ + 3³ = 1 + 125 + 27
- 370 = 3³ + 7³ + 0³ = 27 + 343 + 0
