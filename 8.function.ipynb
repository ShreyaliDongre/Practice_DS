```python
#function:
#a function is a block of code which only runs when it is called. 
# You can pass data, known as parameters, into a function. A function can return data as a result.
#creating a function
#code reliability and reusability
#better code organization
#easier to debug and maintain
#reduce code duplication
#syntax:
#def function_name(parameters):
    #function body
    #return value (optional)
#example:
def greet(name):
    return f"Hello, {name}!"
print(greet("Alice"))
```

```python
#Arguments: an argument is a value that is passed to a function when it is called. There are several types of arguments in Python:
#1. Positional arguments: These are the most common type of arguments. They are passed to a function in the order they are defined. For example:
def greet(name, age):
    print("Name:",name)
    print("Age:",age)
greet("Alice", 30)  # Output: Name: Alice, Age: 30
#2. Keyword arguments: These are arguments that are passed to a function using the name of the parameter. This allows you to specify the value of an argument without having to worry about the order of the parameters. For example:
def greet(name, age):
    print("Name:",name)
    print("Age:",age)
greet(age=30, name="Alice")  # Output: Name: Alice, Age: 30
#3. Default arguments: These are arguments that have a default value assigned to them. If
    # you do not provide a value for a default argument when calling the function, the default value will be used. For example:
def greet(name, age=30):
    print("Name:",name)
    print("Age:",age)
greet("Alice")  # Output: Name: Alice, Age: 30
greet("Bob", 25)  # Output: Name: Bob, Age:

#parameterized function: A parameterized function is a function that takes one or more parameters as input. These parameters can be used within the function to perform specific tasks or calculations. For example:
def calculate_area(radius):
    area = 3.14 * radius ** 2
    return area
#difference between parameterized and argument function:
#1.param- defined in the function definition, 
# argument- value passed to the function when it is called
#2.param- act as placeholders for the values that will be passed to the function, 
# argument- actual values that are passed to the function when it is called
#3.param- used to define the function and specify the input it requires, 
# argument- used to provide the input values when calling the function




def greet(name):# 'name' is a parameter   
    greet("Alice")  # 'Alice' is an argument
```

```python
#recursive function: A recursive function is a function that calls itself in order to solve a problem. 
# It typically has a base case that stops the recursion and a recursive case that breaks the problem into smaller subproblems. For example:

#syntax:
#def recursive_function(parameters):
    #base case
    #recursive case
#Example of recursive function to calculate factorial:
def factorial(n):
    if n == 0:  # Base case
        return 1
    else:       # Recursive case
        return n * factorial(n - 1)
```

```python
#Exception handling in functions:
#Exception handling is a mechanism in Python that allows you to handle errors gracefully without crashing the program. You can use try-except blocks to catch and handle exceptions that may occur during the execution of a function. For example:
def divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."
    
```
